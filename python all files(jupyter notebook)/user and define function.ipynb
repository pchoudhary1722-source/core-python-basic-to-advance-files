# create calculator(a,b,operation) that perfoms +,-,* and / depending on the operation entered

```python
def cal(a,b,operator):
    if operator=="+":
        print(a+b)
    elif operator=="-":
        print(a-b)
    elif operator=="*":
        print(a*b)
    elif operator=="/":
        print(a/b)
    else:
       print("incorrect value")
a = int(input("enter yor number "))
b = int(input("enter ur number "))
operator = input("enter ur input(+,-,*,/)") 
cal(a,b,operator)
```

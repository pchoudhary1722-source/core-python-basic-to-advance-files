# Method Overriding

Meaning: A child class provides its own version of a method that already exists in the parent class.

```python
class Father:
    def work(self):
        print("Father is working")

class Son(Father):
     def work(self):
        print("Son is working")

S=Son()
S.work()
```

# 2. Method Overloading

Meaning: Using the same method name with different numbers/types of arguments.

```python
class Calculator:
    def add(self, a, b):
        return a+b

    def add(self, a, b, c):
         return a + b + c

c=Calculator()
c.add(20,30,40)
```

```python
class Calculator:
    def add(self, a, b, c=0):
         return a + b + c

c=Calculator()
print(c.add(30,40))
print(c.add(40,50,60))
```

```python
#super function
class Father:
   def work(self):
       print("Father is working")

class Son (Father):
    def work(self):
        super().work()
        print("Son is working")

s=Son()
s.work()
```

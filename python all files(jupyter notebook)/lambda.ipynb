# lambda

# lambda: in python a lambda is an anonymous(nameless) function that can have any number of arguments but only one expression. it's often used for short, simple operation where defining a full function with def woubld be unnecessary

# formula of lambda= lambda argument: expression

```python
def squ(x):
    return x*x
```

```python
squ(2)
```

```python
squ= lambda x: x*x
```

```python
squ(45)
```

```python
squ(90)
```

```python
def add(a,b):
    return(a+b)

result= add(12,12)
print("result:",result)
```

# map

```python
number=[1,2,3,4,5]

result=list(map(lambda x: x*2, number))
print(result)
```

```python
words=["hello","bird","python"]

uppercased= list(map(lambda x:x.upper(), words))
print(uppercased)
```

# find the square of a number

```python
squ= lambda x: x**2
print(squ(12))
```

# add two numbers

```python
add= lambda x,b: x+b
print(add(12,34))
```

# find the cube of a number

```python
cube= lambda x: x**3 
print(cube(12))
```

# check whether a number is even or odd

```python
is_even= lambda x:"even" if x%2==0 else "odd"
print(is_even(4))
print(is_even(7))
```

# find the larger of two numbers

```python
larger= lambda x , y: x if x>y else y
print(larger(10,17))
```

# find the smaller of two numbers

```python
smaller= lambda x,y: x if x<y else y
print(smaller(13,89))
```

# check the number is postive, negative , zero

```python
check= lambda x:"postive" if x>0 else"negative" if x<0 else "zero"
print(check(6))
print(check(-1))
print(check(0))
```

# use lambd WHITH MAP() TO DOUBLE EVERY NUMBER

```python
number=[12,13,14,15,15]
double= list(map(lambda x:x*2, number))
print(double)
```

# use lambda with filter()to get only even numbers

```python
number=[12,13,14,15,16,17,18,18,19,20]
evens= list(filter(lambda x: x%2==0, number))
print(evens)
```

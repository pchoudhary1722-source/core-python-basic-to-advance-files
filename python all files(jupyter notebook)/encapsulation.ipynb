# public
public member can be acces from anywhere.
in python, normally we simply write the variable name.

```python
class student:
    def __init__(self):
        self.name="priyanka"

s=student()
print(s.name)
```

```python
s.name="apple"

print(s.name)
```

# private
private members are written with two underscores__

```python
class bank:
    def __init__(self):
        self.__balance=10000

b=bank()
print(b.__balance)
```

```python
class bank:
    def __init__(self):
        self.__balance=10000
   
    def show_balance(self):
        print(self.__balance)

b=bank()
b.show_balance()
```

```python
class bank:
    def __init__(self):
        self.__balance=90000
   
    def show_balance(self):
        print(self.__balance)

b=bank()
b.show_balance()
```

# best easy example

```python
class student:
    def __init__(self):
        self.name="priyanka"
        self._age=27
        self.__marks=78

s=student()
print(s.name)
print(s._age)
print(s.__marks)
```
